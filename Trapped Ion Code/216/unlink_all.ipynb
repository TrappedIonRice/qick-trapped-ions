```python
import os
from pathlib import Path

ROOT  = Path("/home/xilinx/jupyter_notebooks/qick/firmware/projects")
MODE  = "text"    # "text" = turn each link back into a text placeholder (matches the PC)
                  # "delete" = remove the link and leave the path empty
APPLY = True     # False = only list what would change; set True to actually do it

found = 0
for dirpath, dirnames, filenames in os.walk(ROOT, followlinks=False):
    if Path(dirpath).name != "out":
        continue
    for name in dirnames + filenames:          # links can appear in either list
        p = Path(dirpath) / name
        if not p.is_symlink():
            continue
        found += 1
        target = os.readlink(p)                # the link's path text, e.g. ../top/...
        print(f"{'CHANGING' if APPLY else 'would change'}: {p}  ->  {target}")
        if APPLY:
            p.unlink()                         # removes only the link; the target file is untouched
            if MODE == "text":
                p.write_text(target)           # plain file with the same text git uses, no newline

print(f"\n{found} link(s) {'changed' if APPLY else 'found'} (mode: {MODE})")
```
